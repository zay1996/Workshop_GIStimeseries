## 5. Inspect the returned objects

```python
print("Returned keys:", list(results))

pattern = results["pattern"]
patch_size = results["patch_size"]
patch_num = results["patch_num"]
gainloss = results["gainloss"]
increase_decrease = results["increase_decrease"]

print("Pattern shape:", pattern.shape)
print("Pattern dtype:", pattern.dtype)
```

```python
transition_labels = {
    -1: "NoData",
     0: "Stable absence",
     1: "Appearing",
     2: "Merging",
     3: "Filling",
     4: "Expanding",
     5: "Disappearing",
     6: "Splitting",
     7: "Perforating",
     8: "Contracting",
     9: "Stable presence",
}

values, counts = np.unique(pattern, return_counts=True)
for value, count in zip(values, counts):
    print(f"{value:>2}  {transition_labels.get(int(value), 'Unknown'):<18} {count:>10,} pixels")
```
